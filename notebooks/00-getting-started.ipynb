## 2. PyTorch & Hardware Acceleration

Check PyTorch installation and available compute devices.

```python
import torch
import torch.nn as nn

print(f"PyTorch version: {torch.__version__}")
print(f"\nAvailable devices:")
print(f"  CPU: ✓")
print(f"  CUDA (NVIDIA GPU): {torch.cuda.is_available()}")
print(f"  MPS (Apple Silicon): {torch.backends.mps.is_available()}")

# Determine best device
if torch.cuda.is_available():
    device = torch.device('cuda')
    print(f"\n✓ Using CUDA GPU: {torch.cuda.get_device_name(0)}")
elif torch.backends.mps.is_available():
    device = torch.device('mps')
    print(f"\n✓ Using Apple Silicon MPS acceleration")
else:
    device = torch.device('cpu')
    print(f"\n✓ Using CPU (no GPU acceleration)")

print(f"\nDefault device: {device}")
```

## 3. Quick PyTorch Test

Let's create a simple tensor and perform operations to verify everything works.

```python
# Create a random tensor and move it to the device
x = torch.randn(3, 3).to(device)
print("Random tensor:")
print(x)

# Perform a simple operation
y = x @ x.T  # Matrix multiplication
print("\nAfter matrix multiplication (x @ x.T):")
print(y)

print("\n✓ PyTorch operations working correctly!")
```

## 4. Simple Neural Network Example

Create a tiny neural network to verify autograd works.

```python
# Simple 2-layer network
model = nn.Sequential(
    nn.Linear(10, 5),
    nn.ReLU(),
    nn.Linear(5, 2)
).to(device)

# Create dummy input
input_data = torch.randn(1, 10).to(device)

# Forward pass
output = model(input_data)

print(f"Model architecture:")
print(model)
print(f"\nInput shape: {input_data.shape}")
print(f"Output shape: {output.shape}")
print(f"Output: {output}")

print("\n✓ Neural network forward pass successful!")
```

## 5. Visualization Test

Quick plot to verify matplotlib works in notebooks.

```python
# Generate some data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create plot
plt.figure(figsize=(10, 4))
plt.plot(x, y, label='sin(x)')
plt.title('Visualization Test')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("✓ Visualization working!")
```
